### RAG : Retrieval Augmented Generation

Retrieval-Augmented Generation (RAG) is the process of optimizing the output of a large language model, so it references an authoritative knowledge base outside of its training data sources before generating a response.

Large Language Models (LLMs) are trained on vast volumes of data and use billions of parameters to generate original output for tasks like answering questions, translating languages, and completing sentences.

RAG extends the already powerful capabilities of LLMs to specific domains or an organization’s internal knowledge base, all without the need to retrain the model.

It is a cost-effective approach to improving LLM output so it remains relevant, accurate, and useful in various contexts.

### RAG Pipeline (Simple Explanation)

RAG works in **two main pipelines**:

1. **Data Ingestion Pipeline**
   Prepares and stores data for searching

2. **Retrieval Pipeline**
   Retrieves useful information and gives it to the LLM to generate better answers


### 1. Data Ingestion Pipeline

This pipeline prepares the documents before users ask questions.

#### Steps:

##### a) Data Ingestion

Collect data from:

* PDFs
* Websites
* Docs
* Databases
* APIs


##### b) Data Parsing

Convert raw data into clean text.

Example:

* Extract text from PDFs
* Remove unnecessary formatting


##### c) Embeddings

Convert text into numerical vectors (embeddings).

Why?

* So AI can understand semantic meaning
* Similar content gets similar vectors

These embeddings are stored inside a **Vector Store**.


### Vector Store

A database that stores embeddings.

It helps quickly find similar information related to a user query.

Examples:

* FAISS
* ChromaDB
* Pinecone


### 2. Retrieval Pipeline

This pipeline runs when a user asks a question.


#### a) User Query

User asks a question.

Example:

```text id="n3x8qp"
"What is LangGraph?"
```


#### b) Retriever

Retriever searches the vector store to find the most relevant information.

It compares:

* user query embedding
  with
* stored document embeddings


#### c) Relevant Context Retrieved

Important matching chunks are fetched from the vector database.


#### d) LLM

The retrieved context + user query are sent to the LLM.

The LLM now gives:

* more accurate
* more relevant
* domain-specific answers


### Final Output

The user receives a better response grounded in real data instead of only the model’s training data.


### Complete Flow

```text id="u7m2kv"
Documents
   ↓
Data Ingestion
   ↓
Data Parsing
   ↓
Embeddings
   ↓
Vector Store
   ↓
User Query
   ↓
Retriever
   ↓
Relevant Context
   ↓
LLM
   ↓
Final Answer
```
