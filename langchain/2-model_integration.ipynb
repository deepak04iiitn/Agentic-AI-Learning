### Streaming and Batch

### Streaming

Streaming in Agentic AI means the AI sends responses step by step while it is still processing tasks.  

Instead of waiting for the full response, users can see the output appear in real time.  

For example, an AI agent can stream generated text, tool results, or updates as soon as they are available.  

This makes the system feel faster, more interactive, and easier to monitor.

```python
for chunk in model.stream("Write me a 200 words paragraph on Artifical Intelligence"):
    print(chunk.text , end="|" , flush=True)
```

### Batch

Batch processing in Agentic AI means handling a collection of requests together instead of processing them one at a time.  

The AI first gathers all tasks, processes them in parallel, and then returns the final results after everything is completed.  

This approach improves performance, reduces processing costs, and is useful for large-scale tasks where real-time responses are not required.

```python
responses = model.batch([
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
])

for response in responses:
    print(response)
```

```python
responses = model.batch([
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
] , config={
    'max_concurrency': 5,  # Limit to 5 parallel calls
})

for response in responses:
    print(response)
```
